```python
# PHASE 1 — CELL 1
# StreamSentinel V2 — Data Exploration
# Imports and project path setup

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# Find project root safely
# --------------------------------------------------

def find_project_root(start_path: Path) -> Path:
    """
    Find the Fraud-detection-ML-V2 project root
    without depending on the notebook's working directory.
    """
    current = start_path.resolve()

    for path in [current, *current.parents]:
        if (
            (path / "data").is_dir()
            and (path / "notebooks").is_dir()
        ):
            return path

    raise FileNotFoundError(
        "Could not find the Fraud-detection-ML-V2 project root. "
        "Make sure this notebook is inside the project's notebooks folder."
    )


PROJECT_ROOT = find_project_root(Path.cwd())

RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DATA_DIR = PROJECT_ROOT / "data" / "processed"
OUTPUTS_DIR = PROJECT_ROOT / "outputs"
FIGURES_DIR = OUTPUTS_DIR / "figures"
METRICS_DIR = OUTPUTS_DIR / "metrics"
REPORTS_DIR = OUTPUTS_DIR / "reports"

# Create output folders if they don't exist
for folder in [
    PROCESSED_DATA_DIR,
    FIGURES_DIR,
    METRICS_DIR,
    REPORTS_DIR
]:
    folder.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)
print("Raw data folder:", RAW_DATA_DIR)
print("Project root exists:", PROJECT_ROOT.exists())
print("Raw data folder exists:", RAW_DATA_DIR.exists())
```

```python
# PHASE 1 — CELL 2
# Discover dataset files

csv_files = sorted(
    RAW_DATA_DIR.glob("*.csv")
)

print("========== DATASET DISCOVERY ==========")

print(
    "CSV files found:",
    len(csv_files)
)

if not csv_files:
    raise FileNotFoundError(
        f"No CSV files were found in:\n{RAW_DATA_DIR}\n\n"
        "Place the new fraud-detection dataset inside data/raw "
        "and run this cell again."
    )

for index, file_path in enumerate(csv_files, start=1):
    print(f"{index}. {file_path.name}")

print("======================================")
```

```python
# PHASE 1 — CELL 5
# Load and validate the training and test datasets

from pathlib import Path
import pandas as pd


# ==================================================
# 1. Find project root safely
# ==================================================

def find_project_root(start_path: Path) -> Path:
    """
    Find the Fraud-detection-ML-V2 project root.
    """

    current = start_path.resolve()

    for path in [current, *current.parents]:

        if (
            (path / "data").is_dir()
            and (path / "notebooks").is_dir()
        ):
            return path

    raise FileNotFoundError(
        "Could not find the Fraud-detection-ML-V2 project root."
    )


PROJECT_ROOT = find_project_root(Path.cwd())

RAW_DATA_DIR = (
    PROJECT_ROOT
    / "data"
    / "raw"
)


# ==================================================
# 2. Define dataset paths
# ==================================================

TRAIN_DATA_PATH = (
    RAW_DATA_DIR
    / "fraudTrain.csv"
)

TEST_DATA_PATH = (
    RAW_DATA_DIR
    / "fraudTest.csv"
)


# ==================================================
# 3. Verify files exist
# ==================================================

if not TRAIN_DATA_PATH.exists():
    raise FileNotFoundError(
        f"Training dataset not found:\n{TRAIN_DATA_PATH}"
    )

if not TEST_DATA_PATH.exists():
    raise FileNotFoundError(
        f"Test dataset not found:\n{TEST_DATA_PATH}"
    )


# ==================================================
# 4. Load datasets
# ==================================================

train_df = pd.read_csv(
    TRAIN_DATA_PATH
)

test_df = pd.read_csv(
    TEST_DATA_PATH
)


# ==================================================
# 5. Display dataset information
# ==================================================

print("========== DATA LOADED ==========")

print(
    "Training file:",
    TRAIN_DATA_PATH.name
)

print(
    "Training shape:",
    train_df.shape
)

print()

print(
    "Test file:",
    TEST_DATA_PATH.name
)

print(
    "Test shape:",
    test_df.shape
)

print()

print("Training columns:")
print(
    train_df.columns.tolist()
)

print()

print("Test columns:")
print(
    test_df.columns.tolist()
)

print("=================================")
```

```python
print("========== BASIC DATASET VALIDATION ==========")

print("\n--- TRAINING DATASET ---")
print("Shape:", train_df.shape)

print("\nData types:")
print(train_df.dtypes)

print("\nMissing values:")
print(train_df.isnull().sum())

print(
    "\nTotal missing values:",
    int(train_df.isnull().sum().sum())
)

print(
    "Duplicate rows:",
    int(train_df.duplicated().sum())
)

print("\n--- TEST DATASET ---")
print("Shape:", test_df.shape)

print("\nData types:")
print(test_df.dtypes)

print("\nMissing values:")
print(test_df.isnull().sum())

print(
    "\nTotal missing values:",
    int(test_df.isnull().sum().sum())
)

print(
    "Duplicate rows:",
    int(test_df.duplicated().sum())
)

train_columns = train_df.columns.tolist()
test_columns = test_df.columns.tolist()

columns_match = (
    train_columns == test_columns
)

print("\n--- SCHEMA CHECK ---")
print(
    "Training and test columns match:",
    columns_match
)

if not columns_match:
    print("\nTraining-only columns:")
    print(
        sorted(
            set(train_columns) - set(test_columns)
        )
    )

    print("\nTest-only columns:")
    print(
        sorted(
            set(test_columns) - set(train_columns)
        )
    )

print("\n==============================================")
```

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()

while PROJECT_ROOT.name != "Fraud-detection-ML-V2" and PROJECT_ROOT.parent != PROJECT_ROOT:
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = PROJECT_ROOT / "data" / "raw" / "fraudTrain.csv"

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATA_PATH}"
    )

df = pd.read_csv(DATA_PATH)

required_columns = [
    "cc_num",
    "merchant",
    "category",
    "amt",
    "lat",
    "long",
    "merch_lat",
    "merch_long",
    "trans_date_trans_time",
    "is_fraud"
]

missing_required_columns = [
    column
    for column in required_columns
    if column not in df.columns
]

print("========== PROJECT DATA VALIDATION ==========")
print("Dataset:", DATA_PATH.name)
print("Shape:", df.shape)
print("Required columns present:", len(missing_required_columns) == 0)

if missing_required_columns:
    print("Missing columns:", missing_required_columns)
    raise ValueError(
        "Dataset does not match the expected StreamSentinel structure."
    )

print("==============================================")
```

```python
print("========== BASIC DATASET VALIDATION ==========")

print("Shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print(
    "\nTotal missing values:",
    int(df.isnull().sum().sum())
)

print(
    "Duplicate rows:",
    int(df.duplicated().sum())
)

print("==============================================")
```

```python
fraud_counts = df["is_fraud"].value_counts(
    dropna=False
)

fraud_percentages = (
    df["is_fraud"]
    .value_counts(
        normalize=True,
        dropna=False
    )
    .mul(100)
)

print("========== FRAUD DISTRIBUTION ==========")

print("Counts:")
print(fraud_counts)

print("\nPercentages:")
print(fraud_percentages.round(4))

print(
    "\nFraud transactions:",
    int((df["is_fraud"] == 1).sum())
)

print(
    "Legitimate transactions:",
    int((df["is_fraud"] == 0).sum())
)

print("========================================")
```

```python
print("========== ENTITY OVERVIEW ==========")

print(
    "Unique customers:",
    df["cc_num"].nunique()
)

print(
    "Unique merchants:",
    df["merchant"].nunique()
)

print(
    "Unique categories:",
    df["category"].nunique()
)

print("\nTop 10 merchant categories:")
print(
    df["category"]
    .value_counts()
    .head(10)
)

print("=====================================")
```

```python
timestamp_series = pd.to_datetime(
    df["trans_date_trans_time"],
    errors="coerce"
)

invalid_timestamps = int(
    timestamp_series.isna().sum()
)

print("========== TIMESTAMP CHECK ==========")

print(
    "Invalid timestamps:",
    invalid_timestamps
)

print(
    "Earliest transaction:",
    timestamp_series.min()
)

print(
    "Latest transaction:",
    timestamp_series.max()
)

print("====================================")
```

```python
print("========== AMOUNT ANALYSIS ==========")

print(
    df["amt"].describe()
)

print(
    "\nMinimum amount:",
    df["amt"].min()
)

print(
    "Maximum amount:",
    df["amt"].max()
)

print(
    "Mean amount:",
    df["amt"].mean()
)

print(
    "Median amount:",
    df["amt"].median()
)

print("====================================")
```

```python
amount_by_fraud = (
    df.groupby("is_fraud")["amt"]
    .agg(
        [
            "count",
            "mean",
            "median",
            "min",
            "max"
        ]
    )
)

print("========== AMOUNT BY FRAUD LABEL ==========")

print(amount_by_fraud)

print("============================================")
```
