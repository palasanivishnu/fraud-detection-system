````python
def find_values_by_keys(data, target_keys):
    found = {}

    if isinstance(data, dict):
        for key, value in data.items():

            normalized_key = str(
                key
            ).lower()

            for target in target_keys:
                if normalized_key == target.lower():
                    found[key] = value

            nested = find_values_by_keys(
                value,
                target_keys
            )

            found.update(nested)

    elif isinstance(data, list):

        for item in data:
            nested = find_values_by_keys(
                item,
                target_keys
            )

            found.update(nested)

    return found

performance_values = find_values_by_keys(
    phase12_data,
    [
        "mean_ms",
        "median_ms",
        "p95_ms",
        "p99_ms",
        "min_ms",
        "max_ms",
        "throughput_transactions_per_second",
        "throughput_txns_per_sec"
    ]
)

print("PHASE 12 PERFORMANCE VALUES")
print("=" * 60)

if performance_values:
    for key, value in performance_values.items():
        print(f"{key}: {value}")
else:
    print(
        "No performance values could be extracted."
    )
````
