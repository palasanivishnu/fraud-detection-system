```python
from pathlib import Path
import json
import math

current_path = Path.cwd().resolve()

PROJECT_ROOT = None

for path in [current_path] + list(current_path.parents):
    if (path / "src").is_dir() and (path / "models").is_dir():
        PROJECT_ROOT = path
        break

if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "Could not locate the Fraud-detection-ML-V2 project root."
    )

MODELS_DIR = PROJECT_ROOT / "models"

DEPLOYMENT_CONFIG_PATH = (
    MODELS_DIR / "deployment_config.json"
)

if not DEPLOYMENT_CONFIG_PATH.exists():
    raise FileNotFoundError(
        f"Deployment configuration not found: {DEPLOYMENT_CONFIG_PATH}"
    )

with open(
    DEPLOYMENT_CONFIG_PATH,
    "r",
    encoding="utf-8"
) as file:
    deployment_config = json.load(file)

print("DEPLOYMENT CONFIGURATION")
print("=" * 60)
print(json.dumps(
    deployment_config,
    indent=2
))
print("=" * 60)

if not isinstance(deployment_config, dict):
    raise TypeError(
        "deployment_config.json must contain a JSON object."
    )

numeric_values = []

def collect_numeric_values(value, location="root"):
    if isinstance(value, dict):
        for key, child in value.items():
            collect_numeric_values(
                child,
                f"{location}.{key}"
            )

    elif isinstance(value, list):
        for index, child in enumerate(value):
            collect_numeric_values(
                child,
                f"{location}[{index}]"
            )

    elif isinstance(value, (int, float)) and not isinstance(value, bool):
        numeric_values.append(
            (location, value)
        )

collect_numeric_values(
    deployment_config
)

invalid_numeric_values = [
    (location, value)
    for location, value in numeric_values
    if not math.isfinite(value)
]

print("\nNUMERIC CONFIGURATION CHECK")
print("=" * 60)

if invalid_numeric_values:
    for location, value in invalid_numeric_values:
        print(
            f"Invalid value at {location}: {value}"
        )

    raise ValueError(
        "Deployment configuration contains invalid numeric values."
    )

print(
    f"Numeric values checked: {len(numeric_values)}"
)
print("All numeric values are finite.")

print("=" * 60)
print("DEPLOYMENT CONFIGURATION: PASS")
```

```python
import json
import math

DECISION_CONFIG_PATH = (
    MODELS_DIR / "decision_engine_config.json"
)

with open(
    DECISION_CONFIG_PATH,
    "r",
    encoding="utf-8"
) as file:
    decision_config = json.load(file)

print("DECISION ENGINE CONFIGURATION")
print("=" * 60)
print(json.dumps(
    decision_config,
    indent=2
))
print("=" * 60)

if not isinstance(
    decision_config,
    dict
):
    raise TypeError(
        "decision_engine_config.json must contain a JSON object."
    )

def walk_numbers(value, path="root"):
    if isinstance(value, dict):
        for key, child in value.items():
            yield from walk_numbers(
                child,
                f"{path}.{key}"
            )

    elif isinstance(value, list):
        for index, child in enumerate(value):
            yield from walk_numbers(
                child,
                f"{path}[{index}]"
            )

    elif isinstance(value, (int, float)) and not isinstance(value, bool):
        yield path, value

decision_numbers = list(
    walk_numbers(
        decision_config
    )
)

invalid_numbers = [
    item
    for item in decision_numbers
    if not math.isfinite(item[1])
]

if invalid_numbers:
    raise ValueError(
        "Decision Engine configuration contains invalid "
        "numeric values."
    )

decision_text = json.dumps(
    decision_config
).lower()

expected_decisions = [
    "allow",
    "otp",
    "review",
    "block"
]

missing_decisions = [
    decision
    for decision in expected_decisions
    if decision not in decision_text
]

if missing_decisions:
    print(
        "Warning: these decisions were not found explicitly:",
        missing_decisions
    )

print("Decision Engine configuration: PASS")
```

```python
import logging

logger = logging.getLogger(
    "streamsentinel_phase14"
)

def safe_execute(operation_name, function_object):
    try:
        result = function_object()

        logger.info(
            "Operation succeeded: %s",
            operation_name
        )

        return {
            "operation": operation_name,
            "status": "SUCCESS",
            "result": result,
            "error": None
        }

    except Exception as exc:
        logger.exception(
            "Operation failed: %s",
            operation_name
        )

        return {
            "operation": operation_name,
            "status": "ERROR",
            "result": None,
            "error": f"{type(exc).__name__}: {exc}"
        }

success_result = safe_execute(
    "controlled_success_test",
    lambda: 42
)

failure_result = safe_execute(
    "controlled_failure_test",
    lambda: (_ for _ in ()).throw(
        ValueError("Controlled error")
    )
)

print("SUCCESS CASE")
print(success_result)

print("\nERROR CASE")
print(failure_result)

if success_result["status"] != "SUCCESS":
    raise RuntimeError(
        "Controlled success handling failed."
    )

if failure_result["status"] != "ERROR":
    raise RuntimeError(
        "Controlled error handling failed."
    )

print("\nError handling wrapper test: PASS")
```

```python
import re

secret_patterns = [
    re.compile(r"api[_-]?key\s*=\s*['\"][^'\"]+['\"]", re.I),
    re.compile(r"secret[_-]?key\s*=\s*['\"][^'\"]+['\"]", re.I),
    re.compile(r"password\s*=\s*['\"][^'\"]+['\"]", re.I),
    re.compile(r"token\s*=\s*['\"][^'\"]+['\"]", re.I)
]

secret_findings = []

scan_files = []

for filename in required_source_files:
    path = SRC_DIR / filename

    if path.exists():
        scan_files.append(path)

for filename in required_model_files:
    path = MODELS_DIR / filename

    if path.exists():
        scan_files.append(path)

for path in scan_files:
    text = path.read_text(
        encoding="utf-8",
        errors="replace"
    )

    for pattern in secret_patterns:
        if pattern.search(text):
            secret_findings.append(
                str(path.relative_to(PROJECT_ROOT))
            )
            break

print("SECRET SCAN")
print("=" * 60)

if secret_findings:
    print("Potential hard-coded secret patterns found:")
    for finding in secret_findings:
        print(f"- {finding}")
else:
    print("No obvious hard-coded secret patterns detected.")

print("=" * 60)
```

```python
import sys
import subprocess
from pathlib import Path

current_path = Path.cwd().resolve()

PROJECT_ROOT = None

for path in [current_path] + list(current_path.parents):
    if (path / "src").is_dir() and (path / "models").is_dir():
        PROJECT_ROOT = path
        break

if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "Could not locate the Fraud-detection-ML-V2 project root."
    )

required_packages = [
    "numpy",
    "pandas",
    "scikit-learn",
    "xgboost",
    "shap",
    "matplotlib"
]

requirements_path = PROJECT_ROOT / "requirements.txt"

installed_versions = []

for package in required_packages:
    try:
        result = subprocess.run(
            [
                sys.executable,
                "-m",
                "pip",
                "show",
                package
            ],
            capture_output=True,
            text=True,
            check=False
        )

        version = None

        for line in result.stdout.splitlines():
            if line.startswith("Version:"):
                version = line.split(":", 1)[1].strip()
                break

        if version:
            installed_versions.append(
                f"{package}=={version}"
            )
        else:
            print(f"Warning: {package} is not installed.")

    except Exception as exc:
        print(
            f"Warning: could not inspect {package}: "
            f"{type(exc).__name__}: {exc}"
        )

if not installed_versions:
    raise RuntimeError(
        "No required package versions could be detected."
    )

requirements_path.write_text(
    "\n".join(installed_versions) + "\n",
    encoding="utf-8"
)

print("REQUIREMENTS FILE CREATED")
print("=" * 60)
print(f"Path: {requirements_path}")
print("=" * 60)

print(requirements_path.read_text(encoding="utf-8"))
```

```python
from pathlib import Path
import ast
import json
import math

current_path = Path.cwd().resolve()

PROJECT_ROOT = next(
    (
        path
        for path in [current_path] + list(current_path.parents)
        if (path / "src").is_dir() and (path / "models").is_dir()
    ),
    None,
)

if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "Fraud-detection-ML-V2 project root could not be found."
    )

SRC_DIR = PROJECT_ROOT / "src"
MODELS_DIR = PROJECT_ROOT / "models"
OUTPUTS_DIR = PROJECT_ROOT / "outputs"
METRICS_DIR = OUTPUTS_DIR / "metrics"
REPORTS_DIR = OUTPUTS_DIR / "reports"
LOG_DIR = OUTPUTS_DIR / "logs"

required_source_files = [
    "feature_contract.py",
    "ml_inference.py",
    "ml_service.py",
    "rule_engine.py",
    "decision_engine.py",
    "shap_explainer.py",
    "detection_service.py",
]

required_model_files = [
    "xgboost_fraud_detector.json",
    "feature_columns.json",
    "model_info.json",
    "deployment_config.json",
    "deployment_manifest.json",
    "rule_engine_config.json",
    "decision_engine_config.json",
]

source_files_present = all(
    (SRC_DIR / filename).exists()
    for filename in required_source_files
)

model_files_present = all(
    (MODELS_DIR / filename).exists()
    for filename in required_model_files
)

source_syntax_valid = True

for filename in required_source_files:
    path = SRC_DIR / filename

    if not path.exists():
        source_syntax_valid = False
        break

    try:
        ast.parse(
            path.read_text(
                encoding="utf-8",
                errors="replace"
            )
        )
    except SyntaxError:
        source_syntax_valid = False
        break

json_files_valid = True

for filename in required_model_files:
    path = MODELS_DIR / filename

    if not path.exists():
        json_files_valid = False
        break

    try:
        with open(
            path,
            "r",
            encoding="utf-8"
        ) as file:
            json.load(file)
    except Exception:
        json_files_valid = False
        break

requirements_path = PROJECT_ROOT / "requirements.txt"

requirements_present = (
    requirements_path.exists()
    and requirements_path.stat().st_size > 0
)

requirements_text = (
    requirements_path.read_text(
        encoding="utf-8",
        errors="replace"
    )
    if requirements_present
    else ""
)

required_packages = [
    "numpy",
    "pandas",
    "scikit-learn",
    "xgboost",
    "shap"
]

requirements_lower = requirements_text.lower()

required_packages_present = all(
    package.lower() in requirements_lower
    for package in required_packages
)

log_file = LOG_DIR / "phase14_production_hardening.log"

logging_test = log_file.exists()

feature_file = MODELS_DIR / "feature_columns.json"

feature_contract_valid = False

if feature_file.exists():
    try:
        with open(
            feature_file,
            "r",
            encoding="utf-8"
        ) as file:
            feature_data = json.load(file)

        expected_features = [
            "amount",
            "amount_vs_avg_ratio",
            "txn_count_last_5min",
            "time_since_last_txn_sec",
            "distance_from_last_location_km",
            "merchant_category_is_new_for_user"
        ]

        configured_features = feature_data

        if isinstance(feature_data, dict):
            for key in [
                "feature_columns",
                "features",
                "columns"
            ]:
                if key in feature_data:
                    configured_features = feature_data[key]
                    break

        feature_contract_valid = (
            list(configured_features) == expected_features
        )

    except Exception:
        feature_contract_valid = False

model_file = MODELS_DIR / "xgboost_fraud_detector.json"

model_file_valid = (
    model_file.exists()
    and model_file.stat().st_size > 0
)

summary_file = (
    METRICS_DIR /
    "phase14_production_hardening_summary.json"
)

report_file = (
    REPORTS_DIR /
    "phase14_production_hardening_report.json"
)

final_checks = {
    "project_found": PROJECT_ROOT.exists(),
    "source_files_present": source_files_present,
    "model_config_files_present": model_files_present,
    "source_syntax_valid": source_syntax_valid,
    "json_files_valid": json_files_valid,
    "requirements_present": requirements_present,
    "required_packages_present": required_packages_present,
    "logging_test": logging_test,
    "feature_contract_valid": feature_contract_valid,
    "model_file_valid": model_file_valid,
    "summary_saved": summary_file.exists(),
    "report_saved": report_file.exists()
}

overall_verification = all(
    final_checks.values()
)

print("PHASE 14 FINAL STATUS")
print("=" * 60)

for name, status in final_checks.items():
    print(f"{name}: {status}")

print("=" * 60)
print(
    f"Overall verification: {overall_verification}"
)

if overall_verification:
    print("PHASE 14 STATUS: READY")
else:
    print("PHASE 14 STATUS: REVIEW REQUIRED")
```
